```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict
```

```python
class States(TypedDict):
    p1 : str
    p2 : str
    p3 : str
```

```python
#parrallel func 1
def p1(state: States):

    ans = "this is stae"
    return {'p1': ans}

def p2(state: States):

    ans = "this is stae"
    return {'p2': ans}


def p3(state: States):

    ans = "this is stae"
    return {'p3': ans}
```

```python
graph = StateGraph(States)

graph.add_node('p1', p1)
graph.add_node('p2', p2)
graph.add_node('p3', p3)

graph.add_edge(START, 'p1')
graph.add_edge(START, 'p2')
graph.add_edge(START, 'p3')
graph.add_edge('p1', END)
graph.add_edge('p2', END)
graph.add_edge('p3', END)

workflow = graph.compile()
```

```python
result = workflow.invoke({})

result
```
